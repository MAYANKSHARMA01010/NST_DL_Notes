# Activation Functions — A Simple Demo (Student Version)

An **activation function** is the small nonlinear function every neuron applies to its output. Each neuron does two steps:

1. A straight-line calculation: $z = w \cdot x + b$  (weight × input + bias)
2. An activation: $a = f(z)$

We look at four common choices for $f$ — **Linear, Sigmoid, Tanh, and ReLU** — and plot the shape of each.

For every function we implement it **twice**:
- **First with NumPy** — writing out the raw math ourselves.
- **Then with PyTorch** — using the framework's built-in version, which is what you'll actually use when building real networks.

Both should produce the *same* curve — that's how you know your NumPy math was right.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch

# A range of input values (z). We keep both a NumPy version and a PyTorch tensor version.
z    = np.arange(-6, 6, 0.05)      # NumPy array
z_t  = torch.from_numpy(z)         # same values as a PyTorch tensor

# Convert either a NumPy array OR a torch tensor to a plain NumPy array (for plotting)
def to_numpy(a):
    return a.detach().numpy() if hasattr(a, 'detach') else np.asarray(a)

# Tiny helper: draw any activation function over its input range
def plot_activation(f, name, xin=z):
    plt.figure(figsize=(5, 3))
    plt.plot(to_numpy(xin), to_numpy(f(xin)))
    plt.title(name)
    plt.xlabel('z'); plt.ylabel('f(z)')
    plt.axhline(0, color='gray', lw=0.5)
    plt.grid(alpha=0.3)
    plt.show()
```

## 6. Summary

| Function | Range | Used in | Main strength | Main weakness |
|---|---|---|---|---|
| Linear | $(-\infty, \infty)$ | (illustration only) | Simple | Whole network collapses to linear |
| Sigmoid | $(0, 1)$ | Binary-classification output | Reads as a probability | Vanishing gradients |
| Tanh | $(-1, 1)$ | Hidden layers (older nets) | Zero-centered | Vanishing gradients |
| ReLU | $[0, \infty)$ | Hidden layers (default today) | Fast, no saturation for $z>0$ | Dying ReLU |

**Takeaway:** the NumPy and PyTorch versions draw the exact same curves — PyTorch just gives you these as fast, ready-made building blocks (`torch.sigmoid`, `torch.tanh`, `torch.relu`) so you don't re-derive the math every time.

# AND Problem

```python
import torch
import torch.nn as nn
```

```python
# AND gate data
X = torch.tensor([[0, 0],
                  [0, 1],
                  [1, 0],
                  [1, 1]], dtype=torch.float32)

y = torch.tensor([[0], [0], [0], [1]], dtype=torch.float32)
```

```python
# Single perceptron: 2 inputs -> 1 output
model = nn.Linear(2, 1)
```

```python
# Manually set weights and bias for AND
model.weight.data = torch.tensor([[1., 1.]])
model.bias.data = torch.tensor([-1.5])
```

```python
# Forward pass
z = model(X)
z
```

```python
pred = (z>=0).int()
pred
```

```python
print("Predicted:", pred.squeeze().tolist()) #tolist() converts the PyTorch tensor into a regular Python list.
print("Actual:   ", y.squeeze().int().tolist())
```

```python
import torch
from matplotlib.figure import Figure
from IPython.display import display  #used to show the Matplotlib figure inside Jupyter Notebook or Google Colab.
from matplotlib.colors import ListedColormap

# 0 will be red, 1 will be green
my_colors = ListedColormap(['red', 'green'])
x_line = torch.tensor([-0.25, 1.7]) #Creates two x-values for drawing the decision boundary line.
                                   #The line will be drawn from x = -0.2 to x = 1.7.
y_line = 0.5 - x_line

"""
For an AND gate, the line needed to separate (1,1)
 from everything else. For an OR gate, the line needs to
 separate (0,0) from (0,1),(1,0), and (1,1).
 An equation that accomplishes this is x_1 + x_2 = 0.5,
  which translates to x_2 = 0.5 - x_1.
"""

fig = Figure(figsize=(5, 4))
ax = fig.subplots()

ax.scatter(X[:, 0], X[:, 1], c=y, cmap=my_colors)
ax.plot(x_line, y_line)

ax.set(title="OR using SLP", xlabel="x1", ylabel="x2",
       xlim=(-0.3, 1.7), ylim=(-0.3, 1.7))
for i in range(len(X)):
    ax.text(X[i, 0] + 0.03, X[i, 1] + 0.03, f"y={int(y[i])}")
display(fig)
```

# Student Activity 2: XOR problem with SLP

```python
# XOR gate data
X = torch.tensor([[0, 0],
                  [0, 1],
                  [1, 0],
                  [1, 1]], dtype=torch.float32)

y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)
```

```python
# Single perceptron: 2 inputs -> 1 output
model = nn.Linear(2, 1)

# Try OR-like weights
model.weight.data = torch.tensor([[1., 1.]])
model.bias.data = torch.tensor([-0.5])

# Forward pass
z = model(X)
pred = (z >= 0).int()

print("XOR using single perceptron")
print("z values: ", z.squeeze().tolist())
print("Predicted:", pred.squeeze().tolist())
print("Actual:   ", y.squeeze().int().tolist())
```

 # XOR problem with MLP

```python
def hidden_layer_xor_forward(x):
    # Hidden neuron 1: OR
    h_or = torch.sigmoid(20*x[:, 0:1] + 20*x[:, 1:2] - 10)

    # Hidden neuron 2: NAND
    h_nand = torch.sigmoid(-20*x[:, 0:1] - 20*x[:, 1:2] + 30)

    # Output neuron: AND of OR and NAND
    output = torch.sigmoid(20*h_or + 20*h_nand - 30)

    hidden_values = torch.cat([h_or, h_nand], dim=1)
    return output, hidden_values

probs, hidden_values = hidden_layer_xor_forward(X)
preds = (probs >= 0.5).float()

print("Hidden layer values: [OR, NAND]")
print(hidden_values.numpy().round(3))

print("Hidden layer model predictions")
print("probabilities:", probs.squeeze().numpy().round(3).tolist())
print("predictions:  ", preds.squeeze().int().tolist())
print("targets:      ", y.squeeze().int().tolist())
```

```python
from matplotlib.figure import Figure
from IPython.display import display

x_line = torch.tensor([-0.3, 1.7])

# BOUNDARY 1: Derived from Hidden Neuron 1 (OR)
# Equation: 20*x1 + 20*x2 - 10 = 0 --> x2 = 0.5 - x1
y_line1 = 0.5 - x_line

# BOUNDARY 2: Derived from Hidden Neuron 2 (NAND)
# Equation: -20*x1 - 20*x2 + 30 = 0 --> x2 = 1.5 - x1
y_line2 = 1.5 - x_line

fig = Figure(figsize=(5, 4))
ax = fig.subplots()

# Plot the XOR points. (Using coolwarm colormap: 0=Blue, 1=Red)
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=80, edgecolors='black', zorder=3)

# Plot the two decision boundaries
ax.plot(x_line, y_line1, label="Hidden 1 (OR)", linestyle="--", color="blue")
ax.plot(x_line, y_line2, label="Hidden 2 (NAND)", linestyle="--", color="red")

# Label the points with their actual 'y' target
for i in range(len(X)):
    ax.text(X[i, 0] + 0.03, X[i, 1] + 0.03, f"y={int(y[i])}", fontsize=10)

ax.set(title="XOR Decision Boundaries using MLP", xlabel="x1", ylabel="x2",
       xlim=(-0.3, 1.7), ylim=(-0.3, 1.7))
ax.legend()

display(fig)
```

```python
# Step function in MLP

import torch

def step(x):
    return (x >= 0).float()


def hidden_layer_xor_forward(x):
    # Hidden neuron 1: OR
    # w1 = 1, w2 = 1, bias = -0.5
    h_or = step(
        1 * x[:, 0:1] +
        1 * x[:, 1:2] -
        0.5
    )

    # Hidden neuron 2: NAND
    # w1 = -1, w2 = -1, bias = 1.5
    h_nand = step(
        -1 * x[:, 0:1] -
        1 * x[:, 1:2] +
        1.5
    )

    # Output neuron: AND of OR and NAND
    # w1 = 1, w2 = 1, bias = -1.5
    output = step(
        1 * h_or +
        1 * h_nand -
        1.5
    )

    hidden_values = torch.cat([h_or, h_nand], dim=1)

    return output, hidden_values


# Forward pass
preds, hidden_values = hidden_layer_xor_forward(X)

print("Hidden layer values: [OR, NAND]")
print(hidden_values.numpy().astype(int))

print("\nModel predictions:")
print("predictions: ", preds.squeeze().int().tolist())
print("targets:     ", y.squeeze().int().tolist())
```
